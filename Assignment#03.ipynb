```python
even_sum = 0
odd_sum = 0

for i in range(1, 51):

    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

print("Sum of even numbers =", even_sum)
print("Sum of odd numbers =", odd_sum)
```

```python
guests = []

for i in range(10):

    name = input("Enter guest name: ")

    if name in guests:
        print("This guest is already added.")
    else:
        guests.append(name)
        print("Guest added successfully.")

print("\nGuest List:")

for guest in guests:
    print(guest)
```

```python
menu = []

for i in range(5):

    dish = input("Enter dish name: ")

    if dish.lower() == "fish":
        print("Sorry! Fish is not available.")
        continue

    menu.append(dish)

print("\nRestaurant Menu:")

for dish in menu:
    print(dish)
```

```python

guests = 300


discard_menu = ["fish", "qeema", "nihari"]

menu = []

menu_price = {}

menu_price_per_head = {}

for i in range(5):

    dish = input("Enter dish name: ")

    if dish.lower() in discard_menu:
        print("Sorry! This dish is not allowed.")
        continue

    menu.append(dish)

    price = float(input("Enter price of " + dish + ": "))

    menu_price[dish] = price

total_menu_price = 0

for dish in menu:
    total_menu_price = total_menu_price + menu_price[dish]

for dish in menu:

    per_head_cost = menu_price[dish] / guests

    menu_price_per_head[dish] = per_head_cost

print("\n========== COMPLETE MENU ==========")

for dish in menu:
    print(dish, "=", menu_price[dish])

print("\nTotal Menu Price =", total_menu_price)

print("\n========== PER HEAD COST ==========")

for dish in menu:
    print(dish, "=", menu_price_per_head[dish])
    
complete_per_head_cost = total_menu_price / guests

print("\nComplete Menu Per Head Cost =", complete_per_head_cost)
```
