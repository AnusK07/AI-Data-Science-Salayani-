```python

class Product:
    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price


class ShoppingCart:
    def __init__(self):
        self.cart = {}

    def add_item(self, product, quantity):
        if product.product_id in self.cart:
            self.cart[product.product_id][1] += quantity
        else:
            self.cart[product.product_id] = [product, quantity]

        print(product.name, "added to cart successfully!")

    def remove_item(self, product_id):
        if product_id in self.cart:
            del self.cart[product_id]
            print("Product removed from cart!")
        else:
            print("Product not found in cart.")

    def view_cart(self):
        if not self.cart:
            print("\nYour cart is empty!")
            return

        print("\n========== SHOPPING CART ==========")
        for product, quantity in self.cart.values():
            print(
                product.name,
                "| Price:", product.price,
                "| Quantity:", quantity,
                "| Subtotal:", product.price * quantity
            )

        print("===================================")

    def calculate_total(self):
        total = 0

        for product, quantity in self.cart.values():
            total += product.price * quantity

        return total


class ShoppingCartSystem:
    def __init__(self):
        self.products = [
            Product(1, "Laptop", 120000),
            Product(2, "Mobile", 65000),
            Product(3, "Headphones", 5000),
            Product(4, "Keyboard", 3000),
            Product(5, "Mouse", 1500)
        ]

        self.shopping_cart = ShoppingCart()

    def display_products(self):
        print("\n========== AVAILABLE PRODUCTS ==========")

        for product in self.products:
            print(
                "ID:", product.product_id,
                "|", product.name,
                "| Price: Rs.", product.price
            )

    def run(self):
        while True:
            print("\n========== SHOPPING CART SYSTEM ==========")
            print("1. View Products")
            print("2. Add Product to Cart")
            print("3. Remove Product from Cart")
            print("4. View Shopping Cart")
            print("5. Checkout")
            print("6. Exit")

            choice = input("\nEnter your choice: ")

            if choice == "1":
                self.display_products()

            elif choice == "2":
                self.display_products()

                try:
                    product_id = int(input("Enter Product ID: "))
                    quantity = int(input("Enter Quantity: "))

                    product = next(
                        (p for p in self.products
                         if p.product_id == product_id), None
                    )

                    if product and quantity > 0:
                        self.shopping_cart.add_item(product, quantity)
                    else:
                        print("Invalid product ID or quantity!")

                except ValueError:
                    print("Please enter valid numbers!")

            elif choice == "3":
                try:
                    product_id = int(input("Enter Product ID to remove: "))
                    self.shopping_cart.remove_item(product_id)
                except ValueError:
                    print("Invalid Product ID!")

            elif choice == "4":
                self.shopping_cart.view_cart()

            elif choice == "5":
                self.shopping_cart.view_cart()

                total = self.shopping_cart.calculate_total()

                if total > 0:
                    print("\nTotal Amount: Rs.", total)
                    print("Thank you for shopping!")
                    self.shopping_cart.cart.clear()
                else:
                    print("Your cart is empty!")

            elif choice == "6":
                print("Thank you for visiting!")
                break

            else:
                print("Invalid choice! Please try again.")


# Main Program
if __name__ == "__main__":
    system = ShoppingCartSystem()
    system.run()
```
